##### 1. 네이버 뉴스에서 "디지털 마케팅" 키워드로 검색한 뉴스 제목 10개를 수집하세요

```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.common.by import By
import time

# 1. 옵션에 대한 설정
options = Options()
options.add_argument("--disable-gpu")
options.add_argument("--start-maximized")
options.add_argument("--headless")

# 2. 드라이버 설정
service = Service("C:\\Users\\user\\Downloads\\chromedriver\\chromedriver.exe")
url = "https://search.naver.com/search.naver?ssc=tab.news.all&where=news&sm=tab_jum&query=%EB%94%94%EC%A7%80%ED%84%B8+%EB%A7%88%EC%BC%80%ED%8C%85"
driver = webdriver.Chrome(service = service, options = options)
driver.get(url)

headlines = driver.find_elements(By.XPATH, "//span[@class = 'sds-comps-text sds-comps-text-ellipsis sds-comps-text-ellipsis-1 sds-comps-text-type-headline1']")

for idx, headline in enumerate(headlines[:10]):
    print(idx+1, headline.text)
```

```python
## 정답

import requests
from bs4 import BeautifulSoup

query = "디지털 마케팅"
url = f"https://search.naver.com/search.naver?where=news&query={query}"

headers = {
    "User-Agent" : "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36"
}

response = requests.get(url, headers = headers)
soup = BeautifulSoup(response.text, "html.parser")

titles = soup.select("a.ox1N67gcMkHhMh7hbfHi.VZxw34rcBHTWHgERxv0b")

for i, title in enumerate(titles[:10], 1)  :
    print(f"{i}. {title.text.strip()}")
```
