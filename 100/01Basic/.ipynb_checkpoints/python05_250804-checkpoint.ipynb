```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
import time

# 1. 옵션, 서비스에 대한 설정

service = Service(ChromeDriverManager().install())
options = Options()
# options.add_argument("--headless")
options.add_argument("--disable-gpu")
options.add_argument("--start-maximized")
options.add_argument("--window-size=1920x1080")
options.add_argument("--lang=ko_KR")
user_agent = "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36"
options.add_argument(f"--user-agent={user_agent}")

#2. 드라이버에 대한 설정
query = "콘텐츠마케팅"
driver = webdriver.Chrome(service = service, options = options)
url = f"https://terms.naver.com/search.naver?query={query}&searchType=&dicType=&subject="
driver.get(url)
time.sleep(2)

contents = driver.find_elements(By.CSS_SELECTOR, "div.info_area > p.desc.__ellipsis.is-truncated")
for content in contents :
    print(content.text.strip())

time.sleep(2)
driver.quit()
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
import time

# 1. 옵션, 서비스에 대한 설정

service = Service(ChromeDriverManager().install())
options = Options()
# options.add_argument("--headless")
options.add_argument("--disable-gpu")
options.add_argument("--start-maximized")
options.add_argument("--window-size=1920x1080")
options.add_argument("--lang=ko_KR")
user_agent = "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36"
options.add_argument(f"--user-agent={user_agent}")

#2. 드라이버에 대한 설정
driver = webdriver.Chrome(service = service, options = options)
url = "https://korean.visitkorea.or.kr/main/area_list.do?type=Place"
driver.get(url)
# try :btn.동의 누를 수 있게끔 만들어야함
# except :
time.sleep(10)

placenames = driver.find_elements(By.XPATH, "//ul[@id='contentList']//strong")

for placename in placenames :
    print(placename.text.strip())

time.sleep(2)
driver.quit()
```
