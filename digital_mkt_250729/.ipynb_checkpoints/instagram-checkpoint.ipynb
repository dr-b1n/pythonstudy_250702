```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
import time
import pickle #쿠키값을 저장하거나 불러와주는 라이브러리

options = Options()
options.add_argument("--start-maximized") #크롬 브라우저 최적화
options.add_argument("window-size=1920x1080")
options.add_argument("lang=ko_KR")

service = Service(ChromeDriverManager().install())

driver = webdriver.Chrome(service=service, options=options)

url = "https://www.instagram.com/accounts/login"
driver.get(url)

print("수동으로 인증 시작")
time.sleep(30)

with open("instagram_cookies.pkl", "wb") as f :
#값이 들어올 때 바이너리코드 형식으로 값을 저장
    pickle.dump(driver.get_cookies(), f)
    # 알고리즘
    # 무언가의 프로그램이 실행되기 위해서 거쳐지는 절차

print("쿠키가 저장 완료되었습니다")
driver.quit()
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
import time
import pickle

options = Options()
options.add_argument("--start-maximized") #크롬 브라우저 최적화
options.add_argument("window-size=1920x1080")
options.add_argument("lang=ko_KR")

service = Service(ChromeDriverManager().install())

driver = webdriver.Chrome(service=service, options=options)

driver.get("https://www.instagram.com")

#쿠키값 넣고 로그인
with open("instagram_cookies.pkl", "rb") as f :
    cookies = pickle.load(f) 
    ##rb: read binary > 값을 읽어오고 실행시켜라
for cookie in cookies :
    driver.add_cookie(cookie)
    
#쿠키값을 가지고 있는 드라이버가 새로고침 한 상태로 들어가게 해줄 것

driver.refresh() #새로고침
time.sleep(5)

try :
    not_now = driver.find_element(By.XPATH, "//button[text()='나중에 하기']")
    not_now.click()
    time.sleep(2)
except :
    pass

# 나중에 하기 페이지가 뜬다면 버튼 눌러주기


#키워드 검색을 위한 설정
keyword = "넷플릭스"
driver.get(f"https://www.instagram.com/explore/tags/{keyword}")
time.sleep(5)

post_links = set()
for _ in range(3) :
    links = driver.find_elements(By.TAG_NAME, "a")
    for link in links :
        href = link.get_attribute("href")
        if href and "/p/" in href :
            post_links.add(href)
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    time.sleep(3)

result = []

for url in list(post_links)[:10] :
    driver.get(url)
    time.sleep(3)
    caption = ""

    try :
        img_tag = driver.find_element(By.XPATH, "//img[@alt]")
        caption = img_tag.get_attribute("alt").strip()
        result.append({"url": url, "caption": caption})
    except Exception as e :
        print(f"캡션 추출 실패 : {url} | 에러 : {e}")
        continue

driver.quit()

for r in result:
    print(r["url"])
    print(r["caption"])
    print("="*60)
```
