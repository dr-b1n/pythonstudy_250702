#### SQL 학습 전, 반드시 알고 가야하는 개념 정의
1. 데이터 (Data)
2. 데이터베이스 (DB)
3. 스키마 (Schema)
4. 테이블 (Table)
5. SQL

##### 1. 데이터 (Data)
- 수집 가능한 값, 사실, 기록물 등의 원초적인 정보
- 숫자: 23, 5, 9
- 문자: "David", "2025-07-31"
- 상태: True, False
- 데이터 그 자체로는 의미가 부족 > 구조화가 되어야 데이터는 비로소 가치를 갖게 됨

##### 2. 데이터베이스 (DataBase)
- 여러가지 다양한 데이터를 구조적이고 체계적으로 저장 및 관리할 수 있도록 해주는 저장소
- 종류
    - 관계형 데이터베이스 (RDBMS > Oracle, mySQL)
    - 비관계형 데이터베이스(관계를 맺긴 하지만 전통적 방식으로 맺지는 않음) (DBMS > MongoDB)
- 오프라인 가게를 운영하는데 일일 고객 주문 정보, 특정 기간동안 사용한 재고 수량 등

##### 3. 스키마 (Schema)
- 데이터베이스의 구조를 정의하는 것
- 데이터베이스를 어떻게 구조화 할 것인지를 정의해놓은 일종의 설계도
- 스키마는 데이터베이스 내부에 존재하는 논리적 구조

##### 4. 테이블 (Table)
- 관계형 데이터베이스 안에 저장된 실제 데이터의 저장 단위
- 구성 요소
  -열 (Column) : 속성 (ex. 고객의 나이, 성별, 이메일 정보)
  -행 (Row) : 레코드 (ex. 한 명의 고객)

##### 5. SQL (Structured Query Language)
- 관계형 데이터베이스를 정의, 조작, 제어할 수 있도록 도와주는 표준 언어
- 종류
    - DDL(Data Definition Language) : CREATE, DROP, ALTER
    - DML(Data Manipulantion Language) : SELECT, INSERT, UPDATE, DELETE
    - DCL (Data Control Language) : GRANT, REVOKE

##### 6. 지금까지 설명한 내용에 대한 정리 & 요약
- 데이터: 가장 하위 개념, 원초적인 정보 의미
- 테이블: 해당 데이터를 담은 (*구조화) 틀
- 스키마: 이러한 테이블 간 관계를 정의한 설계도면
- 데이터베이스 : 스키마의 개념 속에 존재하는 테이블이 포함된 상위 구조 및 개념
- SQL : 이 모든 구조를 만들고, 조작하고, 제어할 수 있도록 해주는 언어

#### SELECT 문법 :
- LIMIT : 가져올 수 있는 데이터 중 일부만 가져오고 싶을 때
- COUNT(*집계함수) : 데이터의 총 갯수를 확인하고 싶을 때
- DISTINCT : 특정 값을 가져올 때, 중복된 값을 출력하고 싶지 않을 때
- SUM : 해당 컬럼값의 합계
- AVG : 해당 컬럼값의 평균
- MAX : 해당 컬럼값의 최대값
- MIN : 해당 컬럼값의 최소값
- GROUP BY : 특정 컬럼값을 기반으로 그룹핑하기
    - SELECT rating FROM film GROUP BY rating => film이라는 테이블에서 rating값을 찾아오는데, 해당 rating의 값을 그룹화해서 찾아와라
    - SELECT COUNT(*) FROM film GROUP BY rating => 각 rating값의 종류별로 몇 개의 데이터가 있는지 확인
    - SELECT COUNT(*) FROM film WHERE 조건문 GROUP BY rating => 특정 조건에 맞는 데이터 중 rating값의 종류별로 몇 개의 데이터가 있는지 확인
    - ORDER BY : 특정 컬럼값을 기준으로 데이터를 정렬 (*오름차순 = ASC || 내림차순 = DESC)
        - 만약  DESC, ASC 중 그 어떤 값도 입력하지 않는다면 기본 디폴트값은 ASC 정렬로 실행됨
```
SELECT 컬럼명
FROM 테이블명
WHERE 조건절
GROUP BY 컬럼
ORDER BY 컬럼
LIMIT 출력갯수
```

##### Data 대분류 리스트 :
1) 수치형 데이터 : (1, 2, 1.1 등) => 해당 수치값이 각각 고유한 의미를 가짐
2) 범주형 데이터 : (남/여, 미성년/성인 etc.) => 중복값을 제거해야하는 경우가 많음
