```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By 
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.keys import Keys
from urllib.request import urlretrieve # url을 수집해오기 위해 쓰는 모듈
import time
import pyautogui


service = Service("C:\\Users\\user\\Downloads\\chromedriver\\chromedriver.exe")

options = Options()

options.add_argument("--disable-gpu") 
options.add_argument("--start-maximized")
# options.add_argument("--headless")
options.add_argument("window-size=1920x1080") #우리 headless 아니고 진짜 이 창 사이즈 정보값으로 하는 중
# options.add_experimental_option("prefs", prefs)
options.add_argument("user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0")
options.add_argument("lang=ko_KR")

driver = webdriver.Chrome(service = service, options = options) 
driver.get("https://davelee-fun.github.io/")

elements = driver.find_elements(By.CSS_SELECTOR, "div.maxthumb img")
sources = list()

for element in elements :
    sources.append(element.get_attribute("src"))
    print(sources)
    #왜 이렇게 할까? > 각각의 아이템들을 저장하기 편하게 하기 위해

for index, source in enumerate(sources) :
    urlretrieve(source, f"IMG/{str(index)}.{source.split(".")[-1]}")
    
time.sleep(2)

driver.quit()
```

#### HTML 크롤링을 위한 XPATH 기본 포맷 (*형태)
- //tagname[@Attribute="value"]
- / : 절대경로를 의미 예) html 문서 내에서 /html/body/div/a
- // : 상대경로, 문서 내에서 바로 검색을 하라는 의미 예)  //h1 > 문서에서 바로 h1 태그를 찾아와라
- //*[@href] : 문서 내 href라는 속성을 가지고 있는 모든 태그를 다 찾아와라
- //a[@href="https://www.google.com"] : 문서 내 a 태그를 찾아오는데, href 속성값으로 구글 닷컴을 찾아와라
- (//a)[3] : 문서 내 a 태그를 찾아오는데, 3번째 a만 찾아와라 (1번부터 찾아옴)
- (//table)[last]: 문서 내 table 태그를 찾아오는데, 마지막번째 table만 찾아와라
- (//a)[position() < 3] : 문서 내 a 태그를 찾아오는데, 앞에서부터 2개만 찾아와라
- //div[@*] : 문서 내 div 태그를 찾아오는데, 어떤 속성이든 상관없이 속성이 하나라도 있으면 찾아와라

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By 
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.keys import Keys
from urllib.request import urlretrieve # url을 수집해오기 위해 쓰는 모듈
import time
import pyautogui


service = Service("C:\\Users\\user\\Downloads\\chromedriver\\chromedriver.exe")

options = Options()

options.add_argument("--disable-gpu") 
options.add_argument("--start-maximized")
# options.add_argument("--headless")
options.add_argument("window-size=1920x1080") #우리 headless 아니고 진짜 이 창 사이즈 정보값으로 하는 중
# options.add_experimental_option("prefs", prefs)
options.add_argument("user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0")
options.add_argument("lang=ko_KR")

driver = webdriver.Chrome(service = service, options = options) 
driver.get("https://davelee-fun.github.io/")

# element = driver.find_element(By.XPATH, "/html/body/div/div/div/h1") # 절대 경로로 찾아오는 방법
element = driver.find_element(By.XPATH, "//h1") # 문서 내에서 직접 찾아오는 방법
print(element.text)

time.sleep(2)

driver.quit()
```
