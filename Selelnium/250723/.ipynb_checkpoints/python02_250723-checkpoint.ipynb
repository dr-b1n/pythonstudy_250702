```python
# 정답

from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By 
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.keys import Keys
import time
import pyautogui

service = Service("C:\\Users\\user\\Downloads\\chromedriver\\chromedriver.exe")
options = Options()
options.add_argument("--disable-gpu") 
options.add_argument("--start-maximized")
# options.add_argument("--headless")
options.add_argument("window-size=1920x1080") #우리 headless 아니고 진짜 이 창 사이즈 정보값으로 하는 중
# options.add_argument("user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0")
options.add_argument("lang=ko_KR")

driver = webdriver.Chrome(service = service, options = options) 


driver.get("https://divjason.github.io/sellenium-test/")

element = driver.find_element(By.ID, "username")
element.clear()
element.send_keys("error@error.com")

#변수명을 다르게 써도 크게 문제가 생기지 않음!
element = driver.find_element(By.ID, "password")
element.clear()
element.send_keys("1234")

time.sleep(5)

# element.send_keys(Keys.RETURN)
element = driver.find_element(By.CSS_SELECTOR, "input[type='submit']")
#  "input[type'submit']" : input의 옵션이 어마무시하게 많기 때문에 따로 속성값을 설정해줘야함
element.click()

time.sleep(5)

element = driver.find_element(By.CSS_SELECTOR, "div.message")
print(element.text)


driver.quit()
```

#### 다음 웹 사이트에 접속해서 selenium을 활용해 크롤링 해오세요
- https://divjason.github.io/sellenium-test/
- input 태그 요소에 아이디 error@error.com / 비밀번호 1234 입력 후 로그인 버튼 자동으로 클릭하기
- 클릭 이후 이동한 페이지에서 message라는 클래스 속성값을 가지고 있는 요소의 텍스트 정보를 출력해주세요
- 해당 페이지에서 로그아웃 버튼을 클릭 후 이동한 페이지에서 다시 message라는 클래스 속성값을 가지고 있는 요소의 텍스트 정보를 출력해주세요

```python
# 내가 연습한 예제
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By 
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.keys import Keys
import time
import pyautogui

service = Service("C:\\Users\\user\\Downloads\\chromedriver\\chromedriver.exe")
options = Options()
options.add_argument("--disable-gpu") 
options.add_argument("--start-maximized")
# options.add_argument("--headless")
options.add_argument("window-size=1920x1080") #우리 headless 아니고 진짜 이 창 사이즈 정보값으로 하는 중
# options.add_argument("user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0")
options.add_argument("lang=ko_KR")

driver = webdriver.Chrome(service = service, options = options) 


driver.get("https://divjason.github.io/sellenium-test/")

element = driver.find_element(By.ID, "username")
element.clear()
element.send_keys("error@error.com")

#변수명을 다르게 써도 크게 문제가 생기지 않음!
element = driver.find_element(By.ID, "password")
element.clear()
element.send_keys("1234")

time.sleep(5)

# element.send_keys(Keys.RETURN)
element = driver.find_element(By.CSS_SELECTOR, "input[type='submit']")
#  "input[type'submit']" : input의 옵션이 어마무시하게 많기 때문에 따로 속성값을 설정해줘야함
element.click()

time.sleep(5)

element = driver.find_element(By.CSS_SELECTOR, "div.message")
print(element.text)

time.sleep(3)

element = driver.find_element(By.CSS_SELECTOR, "input[type = 'submit']")
element.click()

time.sleep(3)

element = driver.find_element(By.CSS_SELECTOR, "div.message")
print(element.text)

time.sleep(6)

driver.quit()
```
