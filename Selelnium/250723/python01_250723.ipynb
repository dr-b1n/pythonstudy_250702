```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By 
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.chrome.options import Options
import time

service = Service("C:\\Users\\user\\Downloads\\chromedriver\\chromedriver.exe")
options = Options()
options.add_argument("--disable-gpu") 
options.add_argument("--start-maximized")
options.add_argument("--headless")

driver = webdriver.Chrome(service = service, options = options) 

driver.get("https://davelee-fun.github.io/")

/# driver.execute_script("window.scrollTo(0, document.body.scrollHeight)")
# execute(실행한다), script(=JS) => 자바스크립트를 실행해라
# window 객체 > 브라우저 그 자체
height = driver.execute_script("return document.body.scrollHeight")
print("scrollHeight :" , height)
#브라우저의 전체 페이지 높이를 읽어줌

client_height = driver.execute_script("return document.documentElement.clientHeight")
print("clientHeight :" , client_height)
#현재 우리의 브라우저, 스크롤을 내리지 않고 직관적인 브라우저 높이가 clientHeight

offset_height = driver.execute_script("return document.body.offsetHeight")
print("offsetheight :" , offset_height)
# 문서 그 자체를 의미

time.sleep(5)

driver.quit()
```
